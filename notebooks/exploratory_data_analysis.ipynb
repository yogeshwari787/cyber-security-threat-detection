# AI Cybersecurity Threat Detection: Exploratory Data Analysis (EDA)
This notebook demonstrates exploratory data analysis on the benchmark network traffic dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('../data/raw/sample_network_traffic.csv')
print('Dataset Shape:', df.shape)
df.head()
```

```python
# Distribution of Attack Classes
plt.figure(figsize=(8, 4))
df['attack_type'].value_counts().plot(kind='bar', color='steelblue')
plt.title('Distribution of Network Traffic Classes')
plt.xlabel('Traffic Category')
plt.ylabel('Connection Count')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Feature analysis: Connection count vs SYN error rate
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='count', y='serror_rate', hue='attack_type', alpha=0.7)
plt.title('SYN Flood Separation: Count vs SError Rate')
plt.show()
```
